Initializing common variables

```sql
%python
import time
import json
import random
import uuid
import csv
from datetime import datetime
import os

catalog = 'databricks_project'
schema = 'generated_files'
volume = 'files'
output_dir = f"/Volumes/{catalog}/{schema}/{volume}"
```

Generating pseudo real-time events

```sql
%python

for i in range(1):
    event = {
        "event_id": str(uuid.uuid4()),
        "user_id": random.randint(1, 6),
        "action_id": random.randint(0, 2),
        "amount": round(random.uniform(10.0, 500.0), 2),
        "timestamp": datetime.now().isoformat()
    }

    file_name = f"{output_dir}/event_{i}_{uuid.uuid4()}.json"
    with open(file_name, "w") as f:
        json.dump(event, f)
        
    print(f"Generated file: {file_name}")
    
    time.sleep(3)
```

Generating dictionaries with changing data for CDC

```sql
%python
users = [
    {'user_id': 1, 'name': 'Alice', 'changing_value': random.randint(1, 10)},
    {'user_id': 2, 'name': 'Bob', 'changing_value': random.randint(1, 10)},
    {'user_id': 3, 'name': 'Samuel', 'changing_value': random.randint(1, 10)},
    {'user_id': 4, 'name': 'John', 'changing_value': random.randint(1, 10)},
    {'user_id': 5, 'name': 'Sara', 'changing_value': random.randint(1, 10)},
    {'user_id': 6, 'name': 'Nick', 'changing_value': random.randint(1, 10)}
]

file_name = f"{output_dir}/users.csv"
with open(file_name, 'w', newline='') as csvfile:
    fieldnames = ['user_id', 'name', 'changing_value']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(users)

print(f"Generated file: {file_name}")

actions = [
    {'action_id': 0, 'name': 'purchase'},
    {'action_id': 1, 'name': 'refund'},
    {'action_id': 2, 'name': 'return'}
]

file_name = f"{output_dir}/actions.csv"
with open(file_name, 'w', newline='') as csvfile:
    fieldnames = ['action_id', 'name']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(actions)

print(f"Generated file: {file_name}")
```
