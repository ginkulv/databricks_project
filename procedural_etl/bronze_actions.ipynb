```sql
%sql
CREATE TABLE IF NOT EXISTS databricks_project.bronze.actions (
    action_id BIGINT,
    name STRING,
    ingested_at TIMESTAMP
) USING DELTA
```

```sql
%python
from pyspark.sql.types import LongType, StringType, StructField, StructType, IntegerType
from pyspark.sql.functions import current_timestamp

file_dir = f"/Volumes/databricks_project/generated_files/files"
table_name = f"databricks_project.bronze.actions"

customSchema = StructType([
    StructField("action_id", LongType(), True),        
    StructField("name", StringType(), True)
])

df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(customSchema) \
    .load(f"{file_dir}/actions.csv")

df.withColumn("ingested_at", current_timestamp()) \
    .write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{table_name}")
```
